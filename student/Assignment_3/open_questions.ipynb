## 3. Cross-Model Comparison

*Compare all six models: MLP, RNN, LSTM, GRU, BERT, GPT*

- How did **mean-pooled FastText embeddings** limit the MLP compared to sequence-based models?
- What advantage did the LSTM's **sequential processing** provide over the MLP?
- Did **fine-tuned LLMs** (BERT/GPT) outperform classical baselines? Explain the performance gap in terms of pretraining and contextual representations.
- **Rank all six models** by test performance. What architectural or representational factors explain the ranking?

1. Since the mean-pooled FastText embeddings is static which ignores contextual information and sequence structure. It averages the vector into a single vector, which loses the information sequently. It treat each word independently. For example, it may classify the "not bad" as "bad", a positive word into a negative word mistakenly.
2. LSTM can capture the sequence of sentence and the dependence of word by word, which keeps the contextual information in the hidden layers. For example, it can capture "not bad" as a negative expression successfully.
3. Yes, fine-tuned LLMs outperform classical baselines. For pretraining，BERT and GPT are pretrained on large corpora before fine-tuning. During pretraining, they learn rich semantic knowledge, syntactic patterns, and contextual relationships between words. So they have strong language understanding. For contextual representations: instead of static FastText embeddings, LLMs can generate contextualized embeddings. It can capture a word meaning based on surrounding context. For example, it can capture "not bad" as a negative expression successfully. Thus, LLMs can capture sentiment information in details, long sentence and sentence structure. 
4. From test macro F1 scores, RNN (0.6618) < MLP (0.6926) < LSTM (0.7280) < GRU (0.7472) < BERT (0.7968) < GPT (0.8083)

- RNN performs the worst. Although it follows sentences sequentially, it has problem with vanishing gradient and has limited ability to capture long sentences. 
- MLP performs better than RNN but still worse. Since it uses static mean-pooled FastText embeddings and ignores word order, it cannot capture contextual relationships between words. 
- LSTM improves a lot. It has gate mechanism, which can control information flow and capture longer contextual dependencies. This helps improve imbalance of minority class performance and macro F1.
- GRU performs slightly better which also has gating mechanisms but has a simpler structure, which make optimization more stable. It develop higher macro F1 and improve better balance across classes.
- BERT and GPT significantly outperform all classical baselines. Since the large-scale pretraining and contextualized embeddings. LLMs are pretrained on huge corpora and learn lots of semantic and contextual representations before fine-tuning. They can generate contextualized embeddings, meaning the representation of a word depends on its surrounding context.
- Between BERT and GPT, GPT has slightly higher macro F1 in this experiment. Both models capture long-range dependencies and complex sentence structures effectively, which explains the large performance gap compared to classical models.
Overall, the results show that the more the model understands context, the better it performs. Models with pretraining and contextual embeddings clearly have an advantage over classical neural networks.
